```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
text= "i love learning agentic ai. python is good language"

```

```python
token=tokenizer.tokenize(text)
print(token)
```

```python
tokens_ids = tokenizer.convert_tokens_to_ids(token)
print(tokens_ids)
tokenid = [tokenizer.cls_token_id] + tokens_ids + [tokenizer.sep_token_id]
print(tokenid)
```
